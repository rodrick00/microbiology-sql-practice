```python
import sqlite3
```

```python
import pandas as pd 
```

```python
db = sqlite3.connect("microbiome.db")
```

```python
q = "SELECT * FROM microbiome_abundance"
df = pd.read_sql_query(q,db)
df
```

```python
mean_abundance_per_taxon = df.groupby("taxon")["relative_abundance"].mean()
mean_abundance_per_taxon
```

```python
db.close()
```

```python
import matplotlib.pyplot as plt
```

```python
taxa = mean_abundance_per_taxon.index
abundances = mean_abundance_per_taxon.values
```

```python
plt.bar(taxa,abundances,color = ["steelblue", "coral", "seagreen", "goldenrod"])
plt.title("Mean Relative Abundance per Taxon")
plt.xlabel("Taxon")
plt.ylabel("Mean Relative Abundance")
plt.tight_layout()
plt.savefig("microbiome_abundance_barchart.png")
plt.show()
```
