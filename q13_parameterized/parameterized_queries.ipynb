```python
def patient_lookups(patient_id):
    safe_query = "SELECT * FROM isolates WHERE patient_id = ?"
    cursor.execute(safe_query,(patient_id,))
    return cursor.fetchall()
```

```python
safe_result_check = patient_lookups("3001")
for row in safe_result_check:
    print(row)
```
