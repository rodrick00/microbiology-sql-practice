```python
import sqlite3
```

```python
db = sqlite3.connect("antibiogram.db")
```

```python
cursor = db.cursor()
```

```python
query = """
SELECT isolates.species_name,antibiotics.antibiotic_name,zones.zone_diameter
FROM zones
JOIN isolates ON isolates.isolate_id = zones.isolate_id
JOIN antibiotics ON antibiotics.antibiotic_id = zones.antibiotic_id 
WHERE zone_diameter <= 14
"""
```

```python
cursor.execute(query)
```

```python
results = cursor.fetchall()
```

```python
for row in results:
    print(row)
```

```python
db.close()
```
